# NumPy

## O que é o NumPy?

O **NumPy** é uma das bibliotecas mais importantes do ecossistema Python para computação numérica.  
Seu principal recurso é o **array multidimensional**, uma estrutura de dados otimizada para armazenar valores do mesmo tipo e realizar operações matemáticas com alto desempenho.

## Por que usar NumPy?

Em comparação com estruturas nativas do Python, o NumPy oferece:

- melhor desempenho em operações numéricas;
- menor consumo de memória em muitos cenários;
- sintaxe mais simples para cálculos em lote;
- base para outras bibliotecas, como Pandas, SciPy, scikit-learn e muitas ferramentas de IA.

> Observação: para executar este notebook, basta ter o pacote `numpy` instalado.

````python
import numpy as np

print("Versão do NumPy:", np.__version__)
````

# 1. Arrays NumPy versus listas Python

Uma **lista Python** é bastante flexível: pode conter tipos diferentes, crescer dinamicamente e ser usada em muitos contextos gerais de programação.

Já um **array NumPy** foi projetado para **cálculos numéricos**.  
Ele normalmente trabalha com dados **homogêneos**, isto é, elementos de um mesmo tipo, o que favorece desempenho e padronização.

## Quando usar lista?

Use lista quando:

- os dados forem variados;
- você precisar de muita flexibilidade estrutural;
- o foco não for computação numérica.

## Quando usar array NumPy?

Use array NumPy quando:

- os dados forem numéricos;
- você precisar fazer contas com muitos valores;
- quiser trabalhar com vetores, matrizes e estatística;
- precisar de melhor performance.

## Ideia principal

Pense assim:

- **lista Python** → estrutura geral;
- **array NumPy** → estrutura especializada para matemática e análise numérica.

````python
# Exemplo com lista Python
lista_notas = [7.5,8.0,9.2,6.8]
print("Lista Python", lista_notas)
print("Tipo: ", type(lista_notas))
````

````python
# Exemplo com array NumPy
array_np = np.array([7.5,8.0,9.2,6.8])
print("Array Numpy", array_np)
print("Tipo: ",type(array_np))
````

````python
# Comparação simples de comportamento
lista_a = [1,2,3]
lista_b = [4,5,6]

print("Concatenando: ", lista_a + lista_b)

array_a = np.array([1,2,3])
array_b = np.array([4,5,6])

print("Soma: ", array_a + array_b)
````

# 2. Tipos de dados, acesso a elementos e slicing

## Tipos de dados no NumPy

Os arrays NumPy possuem um atributo importante chamado `dtype`, que indica o tipo de dado armazenado.  
Esse controle é relevante porque influencia:

- uso de memória;
- precisão dos cálculos;
- compatibilidade com outras operações.

Exemplos comuns de tipos:

- `int64` → inteiros;
- `float64` → números decimais;
- `bool` → valores lógicos;
- `str_` ou tipos textuais em alguns contextos.

## Acesso por índice

Assim como nas listas Python, o índice começa em `0`.

Exemplo:

- índice `0` → primeiro elemento;
- índice `1` → segundo elemento;
- índice `-1` → último elemento.

## Slicing

O **slicing** permite extrair partes de um array usando a sintaxe:

```python
array[inicio:fim:passo]
```

Onde:

- `inicio` é o índice inicial;
- `fim` é o índice final (não incluído);
- `passo` é o intervalo entre os elementos.

Essa técnica é muito útil para selecionar subconjuntos de dados.

````python
idades = np.array([14,15,16,17,18])
print("Array: ", idades)
print("dTypes: ", idades.dtype)
````

````python
pesos = np.array([5.5,3,5], dtype=str)
print("dTypes: ", pesos.dtype)
print(pesos)
````

````python
valores = np.array([10,20,30,40,50,60])
print("Primeiro elemento: ", valores[0])
print("Segundo elemento: ", valores[1])
print("Terceiro elemento: ", valores[2])
print("Último elemento: ", valores[-1])
print("Do índice 1 ao 3: ", valores[1:4])
print("De 2 em 2: ", valores[::2])
print("Invertido: ", valores[::-1])
````

# 3. Operações aritméticas vetorizadas e desempenho

Uma das maiores vantagens do NumPy é a **vetorização**.  
Isso significa que podemos aplicar operações a todos os elementos de um array de forma direta, sem escrever laços manualmente em muitos casos.

## Operações comuns

Com arrays NumPy, podemos fazer:

- soma: `a + b`
- subtração: `a - b`
- multiplicação: `a * b`
- divisão: `a / b`
- potência: `a ** 2`

Também podemos combinar arrays com números escalares.

## Vantagem prática

Em vez de percorrer elemento por elemento com `for`, o NumPy realiza operações em bloco.  
Isso deixa o código:

- mais curto;
- mais legível;
- geralmente mais rápido.

## Sobre desempenho

Em tarefas numéricas maiores, arrays NumPy tendem a ser mais eficientes que listas Python, principalmente quando há muitos cálculos repetitivos.

````python
lista_grande = list(range(1000000))
````

````python
%timeit lista_grande_quadrado = [x ** 2 for x in lista_grande]
````

````python
import numpy as np
````

````python
lista_grande_np = np.array(range(1000000))
````

````python
%timeit lista_grande_np_quadrado = lista_grande_np ** 2
````

````python
valores = np.array([2,4,6,8])
print("Soma: ",valores + 5)
print("Multiplicação: ", valores * 5)
print("Quadrado: ", valores ** 2)
print("Divisão: ", valores/2)
````

# 4. Arrays multidimensionais e matrizes

Até aqui trabalhamos bastante com arrays unidimensionais.  
Agora vamos avançar para arrays com **mais de uma dimensão**.

## Array unidimensional

Representa uma sequência simples de valores, como um vetor:

```python
[10, 20, 30]
```

## Array bidimensional

Representa uma estrutura em linhas e colunas, como uma tabela ou matriz:

```python
[
  [1, 2, 3],
  [4, 5, 6]
]
```

## Conceitos importantes

- `ndim` → número de dimensões;
- `shape` → formato do array (linhas, colunas, etc.);
- `size` → quantidade total de elementos.

## Acesso em matriz

Em um array bidimensional, acessamos assim:

```python
matriz[linha, coluna]
```

Isso é extremamente útil para planilhas, imagens, tabelas e conjuntos de dados estruturados.

````python
vetor = np.array([10, 20, 30, 40])
````

````python
matriz = np.array([
    [10, 20, 30, 40],
    [1, 2, 3, 4],
    [7, 8, 9, 10]
])
````

````python
print(matriz[0,0])
print(matriz[1,1])
print(matriz[1,3])
print(matriz[1,0])

print(matriz.ndim)
print(matriz.shape)
print(matriz.size)
````

````python
print("Linha 0, coluna 2: ", matriz[0,2])
print("Segunda linha: ", matriz[1])
print("Terceira coluna: ", matriz[:,2])
print("Submatriz 2x2: ", matriz[:2, :2])
````

# 5. Estatística com NumPy

O NumPy possui várias funções estatísticas prontas, muito úteis para análise de dados.

## Máximo e mínimo

- `np.max(array)` → maior valor
- `np.min(array)` → menor valor

## Média

- `np.mean(array)` → média aritmética

## Mediana

- `np.median(array)` → valor central dos dados ordenados

## Amplitude

A amplitude pode ser calculada como:

```python
np.max(array) - np.min(array)
```

Ela mostra a diferença entre o maior e o menor valor.

## Percentis e quartis

- `np.percentile(array, 25)` → 1º quartil
- `np.percentile(array, 50)` → mediana
- `np.percentile(array, 75)` → 3º quartil

## Desvio padrão

- `np.std(array)` → mede a dispersão dos dados em relação à média

Quanto maior o desvio padrão, mais espalhados os dados tendem a estar.

````python
dados = np.aray([10,20,30,40,50])
print("Máximo: ", np.max(dados))
print("Minimo: ", np.min(dados))
````

## 6.1 `np.arange()`

A função `np.arange()` cria uma sequência numérica com início, fim e passo.

### Sintaxe geral

```python
np.arange(inicio, fim, passo)
```

- o valor `fim` **não é incluído**;
- o passo pode ser positivo ou negativo.

Ela é muito útil quando precisamos gerar intervalos regulares.

````python
print(np.arange(1,101,1))
````

## 6.2 `np.linspace()`

A função `np.linspace()` gera uma quantidade definida de valores igualmente espaçados entre dois extremos.

### Sintaxe geral

```python
np.linspace(inicio, fim, quantidade)
```

Diferente de `arange()`, aqui você define **quantos valores quer**, e não o passo diretamente.

````python
print(np.linspace(0,1,10))
````

## 6.3 `np.random.randint()`
