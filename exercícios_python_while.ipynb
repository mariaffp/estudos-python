## Exercício 1: Escreva um programa que imprima os números de 1 a 10 usando while.

```python
x = 1
while x <= 10:
  print(x)
  x += 1
```

## Exercício 7: Calcule o fatorial de um número dado pelo usuário usando while.

```python
num = int(input("digite um numero: "))
fatorial = 1
n = 1
while n <= num:
  fatorial *= n
  n += 1
print(fatorial)
```

## Exercício 11: Leia números até que o usuário digite 0 e calcule a soma deles.

```python
somador = 0
while True:
  x = int(input("digite números para saber sua soma e digite o 0 para parar "))
  somador = somador + x
  if x == 0:
    break
print(somador)
```

## Exercício 12: Leia notas até que o usuário digite -1 e calcule a média.

```python
media = 0
cont = 0
acumulador = 0
while True:
  notas = float(input("digite as notas e digite -1 para sair :"))
  if notas == -1:
    break
  cont = cont + 1
  acumulador = acumulador + notas
media = acumulador/cont
print(media)
```

## Exercício 21: Imprima a tabuada de todos os números de 1 a 5.

```python
x = 1
while x <=5:
  cont = 0  #reposiciona a variável cont para 0 a cada novo x , por isso ela é declarada dentro do while
  print(f"tabuada do {x} : ")
  while cont <= 10:
    resultado = x * cont
    print(resultado)
    cont = cont + 1
  x = x + 1
```

## Exercício 22: Imprima todos os pares (i,j) de 1 a 3.

```python
i = 1
while i < 4:
  j = 1
  while j < 4:
    print([i,j])
    j = j + 1
  i = i + 1
```

## Exercício 23: Gere todos os pares de números de 1 a 10 cuja soma seja par.

```python
i = 1
while i < 11:
    j = 1
    while j < 11:
        if (i + j) % 2 == 0:
            print([i, j])
        j = j + 1
    i = i + 1
```
