# Exercícios sobre Funções em Python

### Exercício 1 – Olá, nome!
Crie uma função que exiba a frase: "Olá, nome". O nome deve ser um parâmetro da função

```python
def exibir(nome):
  print(f"Olá, {nome}")
exibir("Maria")
```

### Exercício 2 – Dobro de um número.
Crie uma função que retorne o dobro de um número. O número deve ser um parâmetro da função

```python
def dobro(n):
  return 2 * n
print(dobro(5))
```

### Exercício 3 – Soma simples
Crie uma função que receba 2 números e retorne a soma entre eles

```python
def soma(a,b):
  return a + b
print(soma(2,3))
```

### Exercício 4 – Par ou ímpar
Crie uma função para retornar se um número é par ou ímpar

```python
def verifica_soma_imp(n):
  if n % 2 == 0:
    return("é par")
  else:
    return("é impar")
print(verifica_soma_imp(9))
```

### Exercício 5 – Média de três números.
Crie uma função que recebe 3 número e retrone a média entre eles

```python

def media(x,y,z):
  return (x + y + z) / 3
print(media(3,9,5))
```

### Exercício 6 – Contar vogais
Crie uma função que retorne a quantidade de vogais de uma frase

```python
def contar_vogais(frase):
  v = 0
  vogais = "aeiou"
  for letra in frase:
    if letra in vogais:
      v+= 1
  return v
print(contar_vogais("Castelo"))
```
